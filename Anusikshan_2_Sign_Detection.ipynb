```python
# =========================
# SIGN LANGUAGE RECOGNITION
# =========================

# Install packages
!pip -q install kagglehub tensorflow

# Imports
import os
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras import layers, models
import kagglehub

# Clear previous graphs (prevents tf.function errors)
tf.keras.backend.clear_session()

# -------------------------
# Download Dataset
# -------------------------
path = kagglehub.dataset_download("grassknoted/asl-alphabet")
print("Dataset downloaded at:", path)

# Correct training folder inside dataset
data_dir = os.path.join(path, "asl_alphabet_train")

# -------------------------
# Data Preprocessing
# -------------------------
datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.2
)

train_data = datagen.flow_from_directory(
    data_dir,
    target_size=(64,64),
    batch_size=32,
    class_mode='categorical',
    subset='training'
)

val_data = datagen.flow_from_directory(
    data_dir,
    target_size=(64,64),
    batch_size=32,
    class_mode='categorical',
    subset='validation'
)

# Number of classes auto-detected
num_classes = len(train_data.class_indices)
print("Classes:", num_classes)

# -------------------------
# CNN Model
# -------------------------
model = models.Sequential([

    layers.Input(shape=(64,64,3)),

    layers.Conv2D(32,(3,3),activation='relu'),
    layers.MaxPooling2D(2,2),

    layers.Conv2D(64,(3,3),activation='relu'),
    layers.MaxPooling2D(2,2),

    layers.Conv2D(128,(3,3),activation='relu'),
    layers.MaxPooling2D(2,2),

    layers.Flatten(),

    layers.Dense(128,activation='relu'),

    layers.Dense(num_classes,activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# -------------------------
# Train Model
# -------------------------
history = model.fit(
    train_data,
    epochs=5,        # increase later if needed
    validation_data=val_data
)

# -------------------------
# Save Model
# -------------------------
model.save("sign_language_model.h5")

print("Training Complete")
```

```python
# working
# ==========================================
# SIGN INPUT -> SENTENCE OUTPUT (ONE CELL)
# ==========================================

!pip -q install kagglehub tensorflow

import os, shutil, numpy as np, tensorflow as tf, kagglehub
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras import layers, models
from tensorflow.keras.preprocessing import image
from IPython.display import Javascript, display
from google.colab.output import eval_js
from base64 import b64decode

tf.keras.backend.clear_session()

# --------------------------------
# 1 Download dataset
# --------------------------------
path = kagglehub.dataset_download("grassknoted/asl-alphabet")
source = os.path.join(path,"asl_alphabet_train","asl_alphabet_train")

# --------------------------------
# 2 Use 5 gesture classes and map them to phrases
# (using existing hand shapes as stand-ins for words)
# --------------------------------
subset_dir="/content/phrase_subset"

if os.path.exists(subset_dir):
    shutil.rmtree(subset_dir)

os.makedirs(subset_dir)

gesture_classes=['A','B','C','D','E']

for g in gesture_classes:
    shutil.copytree(
        os.path.join(source,g),
        os.path.join(subset_dir,g)
    )

# Map gestures to sentence outputs
sentence_map = {
"A":"Hello, how are you?",
"B":"I need help.",
"C":"Thank you very much.",
"D":"Yes, I understand.",
"E":"No, I do not understand."
}

# --------------------------------
# 3 Data
# --------------------------------
datagen=ImageDataGenerator(
rescale=1./255,
validation_split=0.2
)

train=datagen.flow_from_directory(
subset_dir,
target_size=(64,64),
batch_size=32,
subset='training'
)

val=datagen.flow_from_directory(
subset_dir,
target_size=(64,64),
batch_size=32,
subset='validation'
)

# --------------------------------
# 4 Model
# --------------------------------
model=models.Sequential([
layers.Input(shape=(64,64,3)),

layers.Conv2D(32,(3,3),activation='relu'),
layers.MaxPooling2D(),

layers.Conv2D(64,(3,3),activation='relu'),
layers.MaxPooling2D(),

layers.Flatten(),

layers.Dense(128,activation='relu'),

layers.Dense(5,activation='softmax')
])

model.compile(
optimizer='adam',
loss='categorical_crossentropy',
metrics=['accuracy']
)

# Short demo training
model.fit(train,epochs=3,validation_data=val)

# --------------------------------
# 5 Webcam capture
# --------------------------------
def take_photo(filename='photo.jpg',quality=0.8):
    js=Javascript('''
    async function takePhoto(quality){
      const div=document.createElement('div');
      const btn=document.createElement('button');
      btn.textContent='Capture';
      div.appendChild(btn);

      const video=document.createElement('video');
      const stream=
      await navigator.mediaDevices.getUserMedia({video:true});

      document.body.appendChild(div);
      div.appendChild(video);

      video.srcObject=stream;
      await video.play();

      await new Promise(
       (resolve)=>btn.onclick=resolve
      );

      const canvas=document.createElement('canvas');
      canvas.width=video.videoWidth;
      canvas.height=video.videoHeight;

      canvas.getContext('2d').drawImage(video,0,0);

      stream.getVideoTracks()[0].stop();
      div.remove();

      return canvas.toDataURL('image/jpeg',quality);
    }
    ''')
    display(js)

    data=eval_js('takePhoto({})'.format(quality))
    binary=b64decode(data.split(',')[1])

    with open(filename,'wb') as f:
        f.write(binary)

    return filename


# --------------------------------
# 6 Predict sign -> sentence
# --------------------------------
print("Show one learned gesture to webcam.")
photo=take_photo()

img=image.load_img(photo,target_size=(64,64))
img=image.img_to_array(img)/255
img=np.expand_dims(img,0)

pred=model.predict(img)

labels=['A','B','C','D','E']
gesture=labels[np.argmax(pred)]

print("\nDetected Sign:",gesture)
print("Sentence Output:")
print(sentence_map[gesture])
```
