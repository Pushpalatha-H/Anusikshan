```python
# ==========================================
# IPL CRICKET SCORE STATISTICS PROJECT
# (NO INTERNET / NO DATASET DOWNLOAD)
# ==========================================


import pandas as pd
import matplotlib.pyplot as plt


# ------------------------------------------
# CREATE SAMPLE IPL DATASET
# ------------------------------------------
data = {
    "season": [2019, 2019, 2020, 2020, 2021, 2021, 2022, 2022],
    "team1": ["MI", "CSK", "RCB", "MI", "CSK", "RCB", "GT", "RR"],
    "team2": ["CSK", "RCB", "MI", "CSK", "MI", "CSK", "RR", "GT"],
    "winner": ["MI", "CSK", "MI", "MI", "CSK", "RCB", "GT", "RR"],
    "toss_winner": ["MI", "RCB", "MI", "CSK", "CSK", "RCB", "GT", "RR"],
    "toss_decision": ["bat", "field", "bat", "field", "bat", "field", "bat", "field"],
    "venue": [
        "Mumbai", "Chennai", "Bangalore", "Mumbai",
        "Chennai", "Bangalore", "Ahmedabad", "Jaipur"
    ]
}


df = pd.DataFrame(data)


print("IPL Dataset Loaded Successfully!")
print(df)
```

```python
# ------------------------------------------
# MATCHES PER SEASON
# ------------------------------------------
season_matches = df["season"].value_counts().sort_index()


plt.figure()
season_matches.plot(kind="bar")
plt.title("Matches per IPL Season")
plt.xlabel("Season")
plt.ylabel("Number of Matches")
plt.show()
```

```python
# ------------------------------------------
# TOP TEAMS BY WINS
# ------------------------------------------
team_wins = df["winner"].value_counts()


plt.figure()
team_wins.plot(kind="bar")
plt.title("Top IPL Teams by Wins")
plt.xlabel("Team")
plt.ylabel("Wins")
plt.show()
```

```python
# ------------------------------------------
# TOSS DECISION ANALYSIS
# ------------------------------------------
toss_decision = df["toss_decision"].value_counts()


plt.figure()
toss_decision.plot(kind="bar")
plt.title("Toss Decision Analysis")
plt.xlabel("Decision")
plt.ylabel("Count")
plt.show()
```

```python
# VENUE ANALYSIS
# ------------------------------------------
venue_matches = df["venue"].value_counts()


plt.figure()
venue_matches.plot(kind="bar")
plt.title("Matches Played at Venues")
plt.xlabel("Venue")
plt.ylabel("Matches")
plt.show()
```

```python
#TOSS WIN vs MATCH WIN
# ------------------------------------------
df["toss_match_win"] = df["toss_winner"] == df["winner"]
result = df["toss_match_win"].value_counts()


plt.figure()
result.plot(kind="bar")
plt.title("Toss Winner vs Match Winner")
plt.xlabel("Result")
plt.ylabel("Count")
plt.show()
```

```python
# ------------------------------------------
# SUMMARY
# ------------------------------------------
print("\nPROJECT SUMMARY")
print("Total Matches:", len(df))
print("Total Seasons:", df['season'].nunique())
print("Total Teams:", pd.concat([df['team1'], df['team2']]).nunique())
print("Most Successful Team:", df['winner'].value_counts().idxmax())


print("\nIPL Cricket Score Statistics Project Completed Successfully!")
```
