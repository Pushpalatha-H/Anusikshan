```python
from langchain_community.document_loaders import PyPDFLoader

loader=PyPDFLoader("Home Price Prediction Using Machine Learning.pdf")
documents=loader.load()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter=RecursiveCharacterTextSplitter(
chunk_size=500,
chunk_overlap=50
)

docs=splitter.split_documents(documents)
```

```python
from langchain_huggingface.embeddings import HuggingFaceEmbeddings    # This class allows you to use pre-trained models from the Hugging Face ecosystem to convert text into numerical vector representations (embeddings).
from langchain_community.vectorstores import FAISS      #

embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
db = FAISS.from_documents(docs, embeddings)
```

```python
from langchain_huggingface.embeddings import HuggingFaceEmbeddings

embeddings=HuggingFaceEmbeddings(
model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
from langchain_community.vectorstores import FAISS

db=FAISS.from_documents(docs,embeddings)
```

```python
from transformers import pipeline

generator = pipeline(
"text-generation",
model="google/flan-t5-base"
)
```

```python
query="What is home price prediction?"

results=db.similarity_search(query,k=3)

context=results[0].page_content

prompt=f"""
Answer based only on this context:

{context}

Question: {query}
"""

answer=generator(prompt,max_length=200)

print(answer[0]['generated_text'])
```
